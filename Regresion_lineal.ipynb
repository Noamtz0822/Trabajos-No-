```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn import datasets, linear_model, model_selection
#Cargar datos de diabetes
X, y = datasets.load_diabetes(return_X_y=True)
#Imprime la forma de los datos y la primer columna
print(X.shape)
print(X[0])
```

```python
#Extraer datos
X = X[:, 2]
print(X.shape)
#Reshape to a 2D array
X = X.reshape(-1, 1)
print(X.shape)
```

```python
#split the data into training nad testing data
X_train, X_test, y_train, y_test = model_selection.train_test_split(X, y, test_size=0.33)
```

```python
#Create a linear regression model and train it wthin our data
model = linear_model.LinearRegression()
model.fit(X_train, y_train)
```

```python
# Predict using our test data
y_pred = model.predict(X_test)
```

```python
#create a scatter plot
plt.scatter(X_test, y_test, color='black')
#plot prediction
plt.plot(X_test, y_pred, color='blue', linewidth=3)
#add label and title
plt.xlabel('Scaled BMI´s')
plt.ylabel('Disease progression')
plt.title('A graph Plot Showing Diabetes against BMI')
plt.show()
```
